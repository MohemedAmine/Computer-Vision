```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('flower.jpg') # read image
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # convert BGR to RGB
plt.imshow(img)
plt.title("Original Image")
plt.axis('off')
plt.show()
```

```python
# Define 3x3 mean filter kernel
G1 = np.ones((3, 3), np.float32) / 9 

blurred_img = cv2.filter2D(img, -1, G1)

plt.imshow(blurred_img)
plt.title("Blurred Image with Mean Filter (G1)")
plt.axis('off')
plt.show()
```

```python
G2 = np.array([[0, 0, 0],
               [0, 2, 0],
               [0, 0, 0]], np.float32)

bright_img = cv2.filter2D(img, -1, G2)

plt.imshow(bright_img)
plt.title("Image after G2 (Intensity ×2)")
plt.axis('off')
plt.show()
```

```python
# Define sharpening filter kernel
#Enhancing Edges and Details
G3 = np.array([[0, -1, 0],
               [-1, 5, -1],
               [0, -1, 0]], np.float32)

sharp_img = cv2.filter2D(img, -1, G3)

plt.imshow(sharp_img)
plt.title("Sharpened Image (G3)")
plt.axis('off')
plt.show()
```

```python
# Define custom filter kernel G4
G4 = np.array([[-1, -1, 0],
               [-1, 3, 0],
               [0, 0, 0]], np.float32)

filtered_G4 = cv2.filter2D(img, -1, G4)
plt.imshow(filtered_G4)
plt.title("Filtered Image (G4)")
plt.axis('off')
plt.show()


# Add 0.5 to correct dark appearance
filtered_G4_corrected = np.clip(filtered_G4 + 0.5, 0, 1)

plt.imshow(filtered_G4_corrected)
plt.title("Filtered Image (G4 + 0.5 correction)")
plt.axis('off')
plt.show()
```

```python
# Blur once
blur_once = cv2.filter2D(img, -1, G1)

# Blur twice
blur_twice = cv2.filter2D(blur_once, -1, G1)

plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.imshow(blur_once)
plt.title("Once Blurred")

plt.subplot(1,2,2)
plt.imshow(blur_twice)
plt.title("Twice Blurred")
plt.show()
```

```python
sobel_x = np.array([[-1, 0, 1],
                    [-2, 0, 2],
                    [-1, 0, 1]], np.float32)

edges_x = cv2.filter2D(cv2.cvtColor(img, cv2.COLOR_RGB2GRAY), -1, sobel_x)

plt.imshow(edges_x, cmap='gray')
plt.title("Sobel Filter - Horizontal Edges")
plt.axis('off')
plt.show()
```

```python
sobel_y = np.array([[-1, -2, -1],
                    [0,  0,  0],
                    [1,  2,  1]], np.float32)

edges_y = cv2.filter2D(cv2.cvtColor(img, cv2.COLOR_RGB2GRAY), -1, sobel_y)

plt.imshow(edges_y, cmap='gray')
plt.title("Sobel Filter - Vertical Edges")
plt.axis('off')
plt.show()
```
